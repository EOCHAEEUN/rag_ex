MonoRouter 설정을 통해 LangChain `ChatOpenAI` 인스턴스를 생성할 수 있는지 확인하기 위한 테스트용입니다.

````python
import os
import sys

print('현재 작업 디렉터리:', os.getcwd())
print('Python 버전:', sys.version)
````

## 2. 모듈 import 확인

````python
from config.common_config import get_llm

print('get_llm import 성공:', get_llm)
````

## 3. 환경변수 설정 여부 확인

보안을 위해 실제 API 키 값은 출력하지 않고, 설정 여부만 확인합니다.

````python
api_key_exists = bool(os.getenv('MONOROUTER_API_KEY'))
base_url = os.getenv('MONOROUTER_BASE_URL')

print('MONOROUTER_API_KEY 설정 여부:', api_key_exists)
print('MONOROUTER_BASE_URL:', base_url)
````

## 4. LLM 인스턴스 생성 확인

아래 모델명은 실제 MonoRouter에서 사용할 수 있는 모델명으로 바꿔주세요.

````python
MODEL_NAME = "gpt-5.4-mini"

llm = get_llm(
    model=MODEL_NAME,
    temperature=0,
    max_tokens=128,
)

print('LLM 인스턴스 생성 성공')
print(type(llm))
````

## 5. 간단 호출 테스트

정상 연결되면 짧은 응답이 출력됩니다.

````python
response = llm.invoke('한 문장으로 자기소개해줘.')

print(response.content)
````

## 6. 반복 사용 예시

모듈화가 잘 되어 있으면 다른 파일이나 노트북에서도 아래처럼 짧게 사용할 수 있습니다.

````python
from config.common_config import get_llm

test_llm = get_llm(MODEL_NAME)
result = test_llm.invoke('LangChain의 장점을 3줄로 설명해줘.')

print(result.content)
````
